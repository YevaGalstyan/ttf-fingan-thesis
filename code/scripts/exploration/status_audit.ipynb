```python
import duckdb
from pathlib import Path
```

```python
BASE = Path.cwd().parent.parent
DATA_DIR = BASE / "data" / "options"
OUT_DIR = BASE / "output"

STATUS_GLOB   = str(DATA_DIR / "status" / "*.parquet")
```

```python
con = duckdb.connect()
con.execute("SET TimeZone = 'UTC'")
```

**Status schema**

Records whether each contract was open for trading or quoting. No prices, so not used.

```python
con.sql(f"""
    SELECT is_trading, is_quoting, COUNT(*) AS n
    FROM read_parquet('{STATUS_GLOB}')
    GROUP BY 1, 2
    ORDER BY n DESC
""").df()
```
